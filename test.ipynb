```python
uniq_paraules09 = paraules09.drop_duplicates(subset=["paraula", "original", "tuti"])
len(uniq_paraules09)
```

Paraules de dues lletres

```python
abc = 'abcçdefghijklmnopqrstuvwxyz'
```

```python
dues_lletres = pd.DataFrame({
    "data": "2021-10-01",
    "paraula": [i + j for i in abc for j in abc],
    "tuti" : False
})
dues_lletres["original"] = dues_lletres["paraula"]

dues_lletres
```

```python
paraules = pd.concat([uniq_paraules09, dues_lletres])
```

```python
paraules["llargada"] = paraules["paraula"].str.len().astype('Int64')
paraules
```
